```python
# Load data

df = pd.read_csv("DataCoSupplyChainDataset2.csv")
```

### Standardize Columns

```python
# Capitalize all characters
# Replace blanks with '_' and replace '(',')' with blanks

df.columns = [c.upper().replace(' ', '_').replace('(', '').replace(')', '') for c in df.columns]
```

```python
df.info()
```

## Remove Unnecessary Columns for Project

```python
df.drop(columns=['CUSTOMER_PASSWORD', 'CUSTOMER_EMAIL', 'PRODUCT_IMAGE', 'PRODUCT_DESCRIPTION', 'LATITUDE', 'LONGITUDE'], inplace=True)
```

```python
df.columns
```

### Renaming ORDER_ITEM attributes to LINE, since ORDER_ITEM entity will be renamed to LINE

```python
# First, fix that messy Date suffix
df.columns = df.columns.str.replace('_DATEORDERS', '')

# Rename the specific 'ORDER_ITEM' bridge entity to 'LINE'
df.columns = df.columns.str.replace('ORDER_ITEM_', 'LINE_')

# Clean up the 'PROFIT_PER_ORDER' to just 'ORDER_PROFIT' for consistency
df.columns = df.columns.str.replace('ORDER_PROFIT_PER_ORDER', 'ORDER_PROFIT')
```

```python
df.info()
```

```python
df.columns
```

### More Column Name Changes for Better Standardization in ERD Creation

```python
# Mapping the old names to your new standardized ERD names
rename_map = {
    'TYPE': 'PAYMENT_TYPE',
    'SALES': 'LINE_SALES',
    'BENEFIT_PER_ORDER': 'ORDER_BENEFIT',
    'SALES_PER_CUSTOMER': 'CUSTOMER_SALES'
}

df.rename(columns=rename_map, inplace=True)
```

### Replace "EE. UU." values to "United States" in CUSTOMER_COUNTRY column

```python
df['CUSTOMER_COUNTRY'] = df['CUSTOMER_COUNTRY'].replace('EE. UU.', 'United States')
```

```python
df['CUSTOMER_COUNTRY'].value_counts()
```

```python
df.info()
```

### Removing more redundant columns for better ERD modeling
- LINE_CARDPROD_ID and PRODUCT_CARD_ID are identical in every row — they're the same foreign key duplicated. You can drop LINE_CARDPROD_ID and just keep PRODUCT_CARD_ID on the Line entity as the FK to Product.
- CATEGORY_ID and PRODUCT_CATEGORY_ID are also identical in every row. One of them is redundant — PRODUCT_CATEGORY_ID is the more appropriate one to keep on the Product entity, so CATEGORY_ID can be dropped.
- ORDER_CUSTOMER_ID and CUSTOMER_ID are also identical — same story. ORDER_CUSTOMER_ID is the FK on the Order entity, so you can drop the standalone CUSTOMER_ID column since it's already represented.

```python
df.drop(columns=['LINE_CARDPROD_ID', 'CATEGORY_ID', 'ORDER_CUSTOMER_ID'], inplace=True)
```

### Remove any remaining trailing/leading white spaces in string values

```python
# Identify all columns with 'object' (string) data types
string_cols = df.select_dtypes(include=['object']).columns


# Strip leading/trailing whitespace from all string (object) columns
df[string_cols] = df[string_cols].apply(lambda x: x.str.strip())
```

### Fix any remaining latin encoding to utf-8

```python
# Identify all columns with 'object' (string) data types
string_cols = df.select_dtypes(include=['object']).columns

# Apply the repair to string columns
for col in string_cols:
    try:
        df[col] = df[col].str.encode('latin1').str.decode('utf-8')
    except (UnicodeEncodeError, UnicodeDecodeError):
        # If a column doesn't have the error, it might fail to re-encode;
        # we skip those to keep the clean data safe.
        continue
```
